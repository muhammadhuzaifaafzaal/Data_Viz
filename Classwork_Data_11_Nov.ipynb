```python
import pandas as pd
import folium
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Read and fetch file, skipping the metadata rows
De = pd.read_excel('/content/ghgp_data_by_year_2023.xlsx', sheet_name='LDC - Direct Emissions', skiprows=3)

# Describe the cleaned dataset
display(De.describe())
De.info()

# Fill missing numeric values with column means
De_filled = De.fillna(De.mean(numeric_only=True))

# Convert to numpy array correctly
GHG_ARRAY = De_filled.to_numpy()

# Emissions columns are at indices 13 through 25
emissions_only = GHG_ARRAY[:, 13:].astype(float)

# Sum down columns ignoring missing values (using filled data)
column_sums = np.sum(emissions_only, axis=0)

# Check for any remaining missing values
print("Missing values sum:\n", np.sum(De_filled.isna()))

# Define the years in chronological order (columns 13 to 25 are 2023 down to 2011, so we reverse them for chronological trend)
years = np.array([2023, 2022, 2021, 2020, 2019, 2018, 2017, 2016, 2015, 2014, 2013, 2012, 2011])
total_emissions_chrono = column_sums[::-1]

# Create the trend line
plt.figure(figsize=(10, 6))
plt.plot(years, total_emissions_chrono, marker='o', color='b', linestyle='-', linewidth=2)
plt.title('Trend Line of Total Direct Emissions Over Years (LDC)')
plt.xlabel('Year')
plt.ylabel('Total Direct Emissions (Metric Tons CO2e)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.xticks(years)
plt.tight_layout()
plt.show()



#Create a bar chart for total direct emissions against sum of all emissions in each row
cols= [
    "2023 Total reported direct emissions from Local Distribution Companies",
    "2022 Total reported direct emissions from Local Distribution Companies",
    "2021 Total reported direct emissions from Local Distribution Companies",
    "2020 Total reported direct emissions from Local Distribution Companies",
    "2019 Total reported direct emissions from Local Distribution Companies",
    "2018 Total reported direct emissions from Local Distribution Companies",
    "2017 Total reported direct emissions from Local Distribution Companies",
    "2016 Total reported direct emissions from Local Distribution Companies",
    "2015 Total reported direct emissions from Local Distribution Companies",
    "2014 Total reported direct emissions from Local Distribution Companies",
    "2013 Total reported direct emissions from Local Distribution Companies",
    "2012 Total reported direct emissions from Local Distribution Companies",
    "2011 Total reported direct emissions from Local Distribution Companies",

]

#add a new column with total emissions across all years for each facility
De["Total_Emissions_All_Years"] = De[cols].sum(axis=1)
print(De["Total_Emissions_All_Years"])

#Group by state and sum total emissions per state

state_emissions = De.groupby("State where Emissions Occur")["Total_Emissions_All_Years"]

state_emissions = De.groupby("State where Emissions Occur")["Total_Emissions_All_Years"].sum()

state_emissions = De.groupby("State where Emissions Occur")["Total_Emissions_All_Years"].sum().reset_index()

print(state_emissions)



#Sort by emissions in descending order fr better visualization

state_emissions = state_emissions.sort_values(by="Total_Emissions_All_Years", ascending=False)


#Extract data to Numpy Arrays for plotting

state  = state_emissions["State where Emissions Occur"]
emissions = state_emissions["Total_Emissions_All_Years"]


#Create Bar Chart now

plt.figure(figsize=(15,6))
plt.bar(state, emissions, color="Skyblue", edgecolor="Black")
plt.title

#Create a pie chart for how much emissions are from transmission pipelines


```
